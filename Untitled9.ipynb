```python
%%writefile common.py
import os, glob, numpy as np, librosa

SR = 22050
WIN = SR
HOP = SR // 2
N_MFCC = 40
FRAMES = 44
EXT = ('.wav', '.mp3', '.ogg', '.flac', '.m4a', '.aac')

def list_audio(folder):
    return sorted(f for f in glob.glob(folder + '/**/*', recursive=True) if f.lower().endswith(EXT))

def load(path):
    y, _ = librosa.load(path, sr=SR, mono=True)
    return y

def windows(y):
    if len(y) < WIN:
        y = np.pad(y, (0, WIN - len(y)))
    return [y[s:s + WIN] for s in range(0, len(y) - WIN + 1, HOP)]

def is_silent(w):
    return float(np.sqrt(np.mean(w ** 2))) < 0.002

def norm(w):
    return (w / (np.max(np.abs(w)) + 1e-6) * 0.9).astype('float32')

def mfcc(w):
    m = librosa.feature.mfcc(y=w, sr=SR, n_mfcc=N_MFCC, n_fft=2048, hop_length=512)[:, :FRAMES]
    if m.shape[1] < FRAMES:
        m = np.pad(m, ((0, 0), (0, FRAMES - m.shape[1])))
    return m.astype('float32')
```

```python
import os, glob, shutil, zipfile, random, numpy as np, librosa, tensorflow as tf, matplotlib.pyplot as plt
from google.colab import files
from common import *

DOG_DIR = '/content/data/dog'
OTHER_DIR = '/content/data/other'
os.makedirs(DOG_DIR, exist_ok=True)
os.makedirs(OTHER_DIR, exist_ok=True)
random.seed(1); np.random.seed(1); tf.random.set_seed(1)

def upload_to(folder):
    up = files.upload()
    for name in up:
        dst = os.path.join(folder, name)
        shutil.move(name, dst)
        if name.lower().endswith('.zip'):
            with zipfile.ZipFile(dst) as z:
                z.extractall(folder)
            os.remove(dst)
    print(f'{len(list_audio(folder))} audio files now in {folder}')
```

```python
dog_files = list_audio(DOG_DIR)
oth_files = list_audio(OTHER_DIR)
assert len(dog_files) >= 3 and len(oth_files) >= 3, 'Upload at least 3 files in EACH option'

def split(f):
    f = f[:]; random.shuffle(f)
    n = len(f); k = max(1, round(n * 0.15))
    return f[:n - 2 * k], f[n - 2 * k:n - k], f[n - k:]

def get_windows(fl):
    out = []
    for f in fl:
        try:
            ws = windows(load(f))
        except Exception as e:
            print('skip', f, e); continue
        out += [w for w in ws if not is_silent(w)]
    return out

def feats(ws):
    if not ws:
        return np.zeros((0, N_MFCC, FRAMES, 1), 'float32')
    return np.stack([mfcc(norm(w)) for w in ws])[..., None]

def augment(w, bg):
    outs = [
        w + np.random.randn(len(w)) * random.uniform(0.01, 0.05),
        w + random.uniform(0.2, 0.8) * norm(random.choice(bg)),
        librosa.effects.pitch_shift(w, sr=SR, n_steps=random.choice([-2, 2])),
        np.roll(w, random.randint(-SR // 4, SR // 4)),
    ]
    return [norm(o) for o in outs]

dtr, dva, dte = split(dog_files)
otr, ova, ote = split(oth_files)

dog_tr = get_windows(dtr); oth_tr = get_windows(otr)
aug = [a for w in dog_tr for a in augment(norm(w), oth_tr)]
Xtr = np.concatenate([feats(dog_tr + aug), feats(oth_tr)])
ytr = np.array([1] * (len(dog_tr) + len(aug)) + [0] * len(oth_tr))

MEAN, STD = float(Xtr.mean()), float(Xtr.std() + 1e-6)
np.save('norm.npy', np.array([MEAN, STD]))
Xtr = (Xtr - MEAN) / STD

def make(dw, ow):
    X = np.concatenate([feats(dw), feats(ow)])
    y = np.array([1] * len(dw) + [0] * len(ow))
    return (X - MEAN) / STD, y

Xva, yva = make(get_windows(dva), get_windows(ova))
Xte, yte = make(get_windows(dte), get_windows(ote))
for n, y in (('train', ytr), ('val', yva), ('test', yte)):
    print(f'{n}: dog={int((y == 1).sum())}  other={int((y == 0).sum())}')
assert len(yva) > 0 and len(yte) > 0, 'Not enough audio for validation/test - upload more files'
```

```python
from tensorflow.keras import layers

def make_cnn():
    i = tf.keras.Input((N_MFCC, FRAMES, 1))
    x = i
    for f in (16, 32, 64):
        x = layers.Conv2D(f, 3, padding='same', activation='relu')(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D(2)(x)
        x = layers.Dropout(0.25)(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(32, activation='relu')(x)
    x = layers.Dropout(0.4)(x)
    o = layers.Dense(1, activation='sigmoid')(x)
    return tf.keras.Model(i, o)

model = make_cnn()
model.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss='binary_crossentropy', metrics=['accuracy'])
n0, n1 = int((ytr == 0).sum()), int((ytr == 1).sum())
cw = {0: len(ytr) / (2 * n0), 1: len(ytr) / (2 * n1)}
hist = model.fit(Xtr, ytr, validation_data=(Xva, yva), epochs=40, batch_size=32, class_weight=cw,
                 callbacks=[tf.keras.callbacks.EarlyStopping(patience=6, restore_best_weights=True)], verbose=2)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
p = model.predict(Xte, verbose=0)[:, 0]
pred = (p >= 0.5).astype(int)
print('Accuracy :', round(accuracy_score(yte, pred), 3))
print('Precision:', round(precision_score(yte, pred, zero_division=0), 3))
print('Recall   :', round(recall_score(yte, pred, zero_division=0), 3))
print('F1       :', round(f1_score(yte, pred, zero_division=0), 3))
try:
    print('ROC-AUC  :', round(roc_auc_score(yte, p), 3))
except Exception as e:
    print('ROC-AUC  : n/a', e)
ConfusionMatrixDisplay(confusion_matrix(yte, pred), display_labels=['Other', 'Dog']).plot(cmap='YlOrBr')
plt.show()
```

```python
%%writefile app.py
import os, re, csv, json, shutil, time, numpy as np, librosa, tensorflow as tf, tensorflow_hub as hub, gradio as gr
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from common import *

# ---------- models ----------
CNN = tf.keras.models.load_model('canine_cnn.keras')
MEAN, STD = [float(v) for v in np.load('norm.npy')]
try:
    YAM = hub.load('https://tfhub.dev/google/yamnet/1')
except Exception:
    YAM = hub.load('https://www.kaggle.com/models/google/yamnet/tensorFlow2/yamnet/1')
names = [r['display_name'] for r in csv.DictReader(open(YAM.class_map_path().numpy().decode()))]
DOG = [i for i, n in enumerate(names) if n in ('Dog', 'Bark', 'Bow-wow', 'Yip', 'Howl', 'Growling')]
NOT = [i for i, n in enumerate(names) if n in ('Music', 'Musical instrument', 'Singing', 'Speech',
                                                'Vehicle horn, car horn, honking', 'Siren', 'Car alarm')]

# ---------- repellent sound ----------
SR_OUT = 48000
def make_repellent(dur=3.0, f0=20000, f1=23000):
    t = np.arange(int(SR_OUT * dur)) / SR_OUT
    f = f0 + (f1 - f0) * t / dur
    s = np.sin(2 * np.pi * np.cumsum(f) / SR_OUT)
    fade = int(0.05 * SR_OUT)
    env = np.ones_like(s); env[:fade] = np.linspace(0, 1, fade); env[-fade:] = np.linspace(1, 0, fade)
    return (0.8 * s * env).astype('float32')

def make_test_tone(dur=3.0):
    return make_repellent(dur, 800, 1600)      # audible, to check the speaker works

# ---------- detection ----------
MYDOG_ROOT = '/content/data/dog/my_dogs'      # inside the dog folder, so retraining (Cells 6-9) also uses it
os.makedirs(MYDOG_ROOT, exist_ok=True)
PROFILE = {'emb': None}                        # average YAMNet "fingerprint" of your own dog
OWN_THR = 0.85                                 # raise = stricter match, lower = looser match

def unit(v):
    return v / (np.linalg.norm(v) + 1e-9)

def rebuild_profile():
    embs = []
    for f in list_audio(MYDOG_ROOT):
        try:
            ws = [w for w in windows(load(f)) if not is_silent(w)]
            for w in ws:
                wav = librosa.resample(norm(w), orig_sr=SR, target_sr=16000).astype('float32')
                embs.append(YAM(wav)[1].numpy().mean(axis=0))
        except Exception as e:
            print('skip', f, e)
    PROFILE['emb'] = unit(np.mean(embs, axis=0)) if embs else None

def analyze(y, cnn_thr=0.6, need=2, ver_thr=0.3, guard=False):
    ws = [w for w in windows(y) if not is_silent(w)]
    if not ws:
        return None
    X = (np.stack([mfcc(norm(w)) for w in ws])[..., None] - MEAN) / STD
    p = CNN.predict(X, verbose=0)[:, 0]
    dog, oth, sim = [], [], []
    for w in ws:
        wav = librosa.resample(norm(w), orig_sr=SR, target_sr=16000).astype('float32')
        sc, em, _ = YAM(wav)
        s = sc.numpy()
        dog.append(s[:, DOG].max()); oth.append(s[:, NOT].max())
        e = unit(em.numpy().mean(axis=0))
        sim.append(float(e @ PROFILE['emb']) if PROFILE['emb'] is not None else 0.0)
    dog, oth, sim = np.array(dog), np.array(oth), np.array(sim)
    ok = (p >= cnn_thr) & (dog >= ver_thr) & (dog > oth)      # both models must agree
    muted = 0
    if guard and PROFILE['emb'] is not None:                  # skip barks that sound like YOUR dog
        own = sim >= OWN_THR
        muted = int((ok & own).sum())
        ok = ok & ~own
    return dict(p=p, dog=dog, oth=oth, ok=ok, muted=muted, trig=int(ok.sum()) >= int(need))

# ---------- UI helpers ----------
MOCHA, CHAMP, AMBER, DEEP, CREAM = '#5B4632', '#F4D9A0', '#C8792C', '#2a1d12', '#F7EBD3'

def card(trig, title, sub):
    st = ('background:linear-gradient(135deg,#C8792C,#F4D9A0);color:#2a1d12;' if trig
          else 'background:rgba(30,20,12,.55);color:#F4D9A0;border:1px solid rgba(244,217,160,.5);')
    return ("<div style='%spadding:20px 22px;border-radius:18px;font-family:Jost,sans-serif'>"
            "<div style='font-family:Cormorant Garamond,serif;font-size:28px;font-weight:600'>%s</div>"
            "<div style='font-size:11px;letter-spacing:.18em;text-transform:uppercase;opacity:.85;margin-top:6px'>%s</div></div>") % (st, title, sub)

def score_plot(r):
    fig, ax = plt.subplots(figsize=(6, 3))
    fig.patch.set_facecolor('#33261a'); ax.set_facecolor('#1d140c')
    t = np.arange(len(r['p'])) * 0.5
    ax.plot(t, r['p'], color=CHAMP, lw=2, label='CNN')
    ax.plot(t, r['dog'], color=AMBER, ls='--', label='YAMNet dog')
    ax.plot(t, r['oth'], color=CREAM, ls=':', alpha=.7, label='YAMNet music/speech/horn')
    ax.set_ylim(0, 1.05); ax.set_xlabel('time (s)', color=CREAM); ax.set_ylabel('score', color=CREAM)
    ax.tick_params(colors=CREAM)
    for sp in ax.spines.values(): sp.set_color('#6b5238')
    ax.legend(facecolor='#33261a', labelcolor=CREAM, fontsize=8, edgecolor='#6b5238')
    fig.tight_layout()
    return fig

def spec_plot(sig):
    fig, ax = plt.subplots(figsize=(6, 3))
    fig.patch.set_facecolor('#33261a'); ax.set_facecolor('#1d140c')
    ax.specgram(sig, Fs=SR_OUT, NFFT=256, noverlap=192, cmap='afmhot', vmin=-100, vmax=-10)
    ax.set_ylim(0, 24000); ax.set_xlabel('time (s)', color=CREAM); ax.set_ylabel('Hz', color=CREAM)
    ax.tick_params(colors=CREAM)
    fig.tight_layout()
    return fig

def to_audio(sig):
    return (SR_OUT, (sig * 32767).astype(np.int16))

def run_file(path, cnn_thr, need, ver_thr, test_tone, guard):
    if not path:
        return card(False, 'No audio', 'Upload or record a clip first.'), None, None, None
    y, _ = librosa.load(path, sr=SR, mono=True)
    r = analyze(y, cnn_thr, need, ver_thr, guard)
    if r is None:
        return card(False, 'All clear', 'Clip is silent.'), None, None, None
    if r['trig']:
        sig = make_test_tone() if test_tone else make_repellent()
        return (card(True, 'Dog bark detected - repellent ON',
                     f"{int(r['ok'].sum())} of {len(r['ok'])} windows confirmed by CNN + YAMNet"),
                score_plot(r), to_audio(sig), spec_plot(sig))
    if r['muted']:
        return (card(False, 'Your own dog - repellent skipped',
                     f"{r['muted']} window(s) matched your My Dog profile"), score_plot(r), None, None)
    return (card(False, 'All clear - not a dog bark',
                 f"{int(r['ok'].sum())} of {len(r['ok'])} windows confirmed (need {int(need)})"),
            score_plot(r), None, None)

LAST = {'t': 0.0, 'a': 0.0}

def live_inner(chunk, buf, cnn_thr, need, ver_thr, test_tone, guard):
    if chunk is None:
        return buf, gr.update(), None
    sr, d = chunk
    was_int = np.issubdtype(d.dtype, np.integer)
    d = d.astype('float32')
    if was_int:
        d = d / 32768.0
    if d.ndim > 1:
        d = d.mean(axis=1)
    d = librosa.resample(d, orig_sr=sr, target_sr=SR)
    buf = np.concatenate([buf, d])[-2 * SR:]
    if len(buf) < 2 * SR:
        return buf, card(False, 'Listening...', 'Collecting audio'), None
    now = time.time()
    if now - LAST['t'] < 5:
        return buf, card(False, 'Cooling down...', 'Repellent just played'), None
    if now - LAST['a'] < 1.0:                  # analyse about once per second, so it never falls behind
        return buf, gr.update(), None
    LAST['a'] = now
    level = float(np.sqrt(np.mean(buf ** 2)))
    if level < 0.002:
        return buf, card(False, 'Listening...', 'Too quiet - check mic allowed and volume'), None
    r = analyze(buf, cnn_thr, need, ver_thr, guard)
    if r is None:
        return buf, card(False, 'Listening...', 'Silence'), None
    info = f"CNN {r['p'].max():.2f} | YAMNet dog {r['dog'].max():.2f} | level {level:.3f}"
    if r['trig']:
        LAST['t'] = now
        sig = make_test_tone() if test_tone else make_repellent()
        return np.zeros(0, 'float32'), card(True, 'Dog bark detected - repellent ON', info), to_audio(sig)
    if r['muted']:
        return buf, card(False, 'Your own dog - repellent skipped', info), None
    return buf, card(False, 'Listening...', 'No dog bark | ' + info), None

def live(chunk, buf, cnn_thr, need, ver_thr, test_tone, guard):
    try:
        return live_inner(chunk, buf, cnn_thr, need, ver_thr, test_tone, guard)
    except Exception as e:                      # never let the stream die silently
        return buf, card(False, 'Live Guard error', str(e)[:120]), None

# ---------- My Dog (personal dog data) ----------
def safe(name):
    return re.sub(r'[^A-Za-z0-9_-]+', '_', name.strip())

def dog_rows():
    rows = []
    for d in sorted(os.listdir(MYDOG_ROOT)):
        p = os.path.join(MYDOG_ROOT, d)
        if os.path.isdir(p):
            info = {}
            if os.path.exists(os.path.join(p, 'profile.json')):
                info = json.load(open(os.path.join(p, 'profile.json')))
            rows.append([info.get('name', d), info.get('breed', ''), info.get('age', ''), len(list_audio(p))])
    return rows

def save_dog(name, breed, age, files):
    if not name or not name.strip():
        return 'Enter your dog\'s name first.', dog_rows()
    if not files:
        return 'Upload at least one bark clip.', dog_rows()
    d = os.path.join(MYDOG_ROOT, safe(name))
    os.makedirs(d, exist_ok=True)
    n = 0
    for f in files:
        src = f if isinstance(f, str) else f.name
        shutil.copy(src, os.path.join(d, os.path.basename(src))); n += 1
    json.dump({'name': name.strip(), 'breed': breed or '', 'age': age or ''}, open(os.path.join(d, 'profile.json'), 'w'))
    rebuild_profile()
    return (f'Saved {n} clip(s) for {name.strip()}. Tick "Don\'t repel my own dog" in Analyze or Live Guard to use it. '
            f'To also train the CNN on it, re-run Cells 6 to 9.'), dog_rows()

def remove_dog(name):
    if not name or not name.strip():
        return 'Type the dog name to remove.', dog_rows()
    shutil.rmtree(os.path.join(MYDOG_ROOT, safe(name)), ignore_errors=True)
    rebuild_profile()
    return f'Removed {name.strip()}.', dog_rows()

rebuild_profile()

CSS = """
@import url('https://fonts.googleapis.com/css2?family=Cormorant+Garamond:ital,wght@0,400;0,500;0,600;1,500&family=Jost:wght@300;400;500&display=swap');
body, .gradio-container {
  background:
    conic-gradient(from 200deg at 92% -8%, transparent 0deg, rgba(255,226,160,.20) 8deg, transparent 16deg,
                   rgba(255,226,160,.12) 24deg, transparent 34deg, rgba(255,226,160,.18) 44deg, transparent 56deg),
    radial-gradient(900px 560px at 90% 0%, rgba(244,200,120,.45), transparent 62%),
    linear-gradient(135deg, #6f5b40 0%, #5a4733 45%, #2f2418 100%) fixed !important;
  font-family: 'Jost', sans-serif !important; color: #F7EBD3 !important;
}
.block, .form, .panel {background: rgba(26,17,10,.55) !important; border: 1px solid rgba(244,217,160,.25) !important;
  border-radius: 16px !important; backdrop-filter: blur(6px);}
label > span, .block-label, .block-title {font-family: 'Jost', sans-serif !important; letter-spacing: .14em;
  text-transform: uppercase; font-size: 11px !important; color: #F4D9A0 !important;}
button[role="tab"] {font-family: 'Jost', sans-serif !important; letter-spacing: .22em; text-transform: uppercase;
  font-size: 12px !important; color: #cdb88d !important; background: transparent !important; border: none !important;}
button[role="tab"][aria-selected="true"] {color: #F4D9A0 !important; border-bottom: 2px solid #F4D9A0 !important;}
button.primary {background: #5B4632 !important; color: #F4D9A0 !important; border: 1px solid #F4D9A0 !important;
  border-radius: 999px !important; letter-spacing: .22em; text-transform: uppercase; font-weight: 500 !important;}
button.primary:hover {background: #F4D9A0 !important; color: #2a1d12 !important;}
:root, .dark, .gradio-container {--slider-color: #5B4632 !important; --color-accent: #5B4632 !important;
  --color-accent-soft: #5B4632 !important; --input-background-fill: #5B4632 !important;}
input[type=range] {accent-color: #5B4632 !important;}
input[type=number], .gradio-container input[type=number] {background: #5B4632 !important; color: #F7EBD3 !important;
  border: 1px solid #F4D9A0 !important;}
input[type=checkbox] {accent-color: #5B4632 !important; background-color: #5B4632 !important; border-color: #F4D9A0 !important;}
input[type=checkbox]:checked {background-color: #5B4632 !important;}
.reset-button, button.reset-button {background: #5B4632 !important; color: #F4D9A0 !important;}
.prose p, .prose strong {color: #F7EBD3 !important;}
table, .table-wrap, .cell-wrap, th, td {background: #3a2b1c !important; color: #F7EBD3 !important;}
th {color: #F4D9A0 !important; letter-spacing: .12em; text-transform: uppercase; font-size: 11px !important;}
textarea, input[type=text] {background: #5B4632 !important; color: #F7EBD3 !important; border: 1px solid #F4D9A0 !important;}
.nav {display:flex; gap:26px; align-items:center; padding:10px 4px; border-bottom:1px solid rgba(244,217,160,.25);
  font-family:'Jost',sans-serif; letter-spacing:.22em; font-size:10px; color:#F4D9A0;}
.nav .dot {width:6px; height:6px; border-radius:50%; background:#F4D9A0; display:inline-block;}
.hero {display:flex; justify-content:space-between; align-items:center; gap:24px; flex-wrap:wrap; padding:20px 6px 10px;}
.eyebrow, .tag {font-family:'Jost',sans-serif; letter-spacing:.22em; font-size:10px; color:#F4D9A0; opacity:.9;}
.title {font-family:'Cormorant Garamond',serif; font-weight:500; font-size:clamp(44px,8vw,84px); line-height:.95;
  color:#F4D9A0; margin:8px 0;}
.title em {color:#C8792C; font-style:italic;}
.pill {display:inline-block; border:1px solid rgba(244,217,160,.6); border-radius:999px; padding:9px 18px;
  margin:12px 8px 0 0; font-family:'Jost',sans-serif; letter-spacing:.22em; font-size:10px; color:#F4D9A0;}
"""

HEADER = """
<div class='nav'><span class='dot'></span><span>HOME</span><span>ANALYZE</span><span>LIVE GUARD</span><span>MY DOG</span><span>ABOUT</span></div>
<div class='hero'>
  <div>
    <div class='eyebrow'>A SAFER WALK<br>EVERY DAY</div>
    <div class='title'>Smart Canine<br><em>Repeller</em></div>
    <div class='tag'>AN AI THAT HEARS A BARK AND ANSWERS WITH A SOUND ONLY DOGS CAN HEAR</div>
    <span class='pill'>LISTEN</span><span class='pill'>DECIDE</span><span class='pill'>REPEL &rarr;</span>
  </div>
</div>
"""

MOCHA_HUE = gr.themes.Color(c50='#f6efe6', c100='#e9dcc8', c200='#d3bc98', c300='#b8956a', c400='#8f6b48',
                            c500='#6f5238', c600='#5B4632', c700='#4a3827', c800='#3a2b1c', c900='#2a1d12', c950='#1d140c')
THEME = gr.themes.Base(primary_hue=MOCHA_HUE, secondary_hue=MOCHA_HUE, neutral_hue=MOCHA_HUE,
                       font=[gr.themes.GoogleFont('Jost'), 'sans-serif']).set(
    body_background_fill='#5a4733', body_background_fill_dark='#5a4733',
    body_text_color='#F7EBD3', body_text_color_dark='#F7EBD3',
    background_fill_primary='#3a2b1c', background_fill_primary_dark='#3a2b1c',
    background_fill_secondary='#4a3825', background_fill_secondary_dark='#4a3825',
    block_background_fill='#3a2b1c', block_background_fill_dark='#3a2b1c',
    block_border_color='#8a7048', block_border_color_dark='#8a7048',
    block_label_text_color='#F4D9A0', block_label_text_color_dark='#F4D9A0',
    block_title_text_color='#F4D9A0', block_title_text_color_dark='#F4D9A0',
    input_background_fill='#5B4632', input_background_fill_dark='#5B4632',
    input_border_color='#F4D9A0', input_border_color_dark='#F4D9A0',
    checkbox_background_color_selected='#5B4632', checkbox_background_color_selected_dark='#5B4632',
    checkbox_border_color_selected='#F4D9A0', checkbox_border_color_selected_dark='#F4D9A0',
    button_primary_background_fill='#5B4632', button_primary_background_fill_dark='#5B4632',
    button_primary_background_fill_hover='#F4D9A0', button_primary_background_fill_hover_dark='#F4D9A0',
    button_primary_text_color='#F4D9A0', button_primary_text_color_dark='#F4D9A0',
    button_primary_text_color_hover='#2a1d12', button_primary_text_color_hover_dark='#2a1d12',
    button_primary_border_color='#F4D9A0', button_primary_border_color_dark='#F4D9A0',
    slider_color='#5B4632', slider_color_dark='#5B4632')

def controls(cnn_default=0.6):
    c = gr.Slider(0.1, 0.95, cnn_default, step=0.05, label='Detection threshold (CNN)')
    n = gr.Slider(1, 6, 2, step=1, label='Confirmation windows')
    v = gr.Slider(0.1, 0.9, 0.3, step=0.05, label='Verifier strictness (YAMNet)')
    t = gr.Checkbox(False, label='Test tone (audible) instead of ultrasonic')
    g = gr.Checkbox(False, label="Don't repel my own dog (uses My Dog data)")
    return c, n, v, t, g

with gr.Blocks(css=CSS, theme=THEME, title='Smart Canine Repeller') as demo:
    gr.HTML(HEADER)
    with gr.Tabs():
        with gr.Tab('Analyze Audio'):
            with gr.Row():
                with gr.Column():
                    a_in = gr.Audio(sources=['upload', 'microphone'], type='filepath', label='Audio clip')
                    a_c, a_n, a_v, a_t, a_g = controls()
                    a_btn = gr.Button('Analyze & Trigger  →', variant='primary')
                with gr.Column():
                    a_card = gr.HTML()
                    a_plot = gr.Plot(label='Scores over time')
                    a_aud = gr.Audio(autoplay=True, label='Repellent', type='numpy')
                    a_spec = gr.Plot(label='Ultrasonic spectrogram')
            a_btn.click(run_file, [a_in, a_c, a_n, a_v, a_t, a_g], [a_card, a_plot, a_aud, a_spec])
        with gr.Tab('Live Guard'):
            with gr.Row():
                with gr.Column():
                    mic = gr.Audio(sources=['microphone'], streaming=True, type='numpy', label='Microphone')
                    l_c, l_n, l_v, l_t, l_g = controls(0.3)
                with gr.Column():
                    l_card = gr.HTML(card(False, 'Listening...', 'Press Record, then allow the microphone'))
                    l_aud = gr.Audio(autoplay=True, label='Repellent', type='numpy')
            buf = gr.State(np.zeros(0, 'float32'))
            mic.stream(live, [mic, buf, l_c, l_n, l_v, l_t, l_g], [buf, l_card, l_aud],
                       stream_every=1, time_limit=3600, show_progress='hidden')
        with gr.Tab('My Dog'):
            gr.Markdown("**Add your own dog's data.** Upload clips of your dog barking. They are saved with the training data "
                        "and let the app skip your own dog.")
            with gr.Row():
                with gr.Column():
                    d_name = gr.Textbox(label='Dog name', placeholder='e.g. Bruno')
                    d_breed = gr.Textbox(label='Breed (optional)')
                    d_age = gr.Textbox(label='Age (optional)')
                    d_files = gr.File(label='Bark clips of your dog', file_count='multiple', type='filepath',
                                      file_types=['audio'])
                    d_save = gr.Button('Save my dog  →', variant='primary')
                with gr.Column():
                    d_msg = gr.Markdown('')
                    d_table = gr.Dataframe(value=dog_rows(), headers=['Name', 'Breed', 'Age', 'Clips'],
                                           interactive=False, label='Saved dogs')
                    d_rm_name = gr.Textbox(label='Remove a dog (type its name)')
                    d_rm = gr.Button('Remove', variant='primary')
            d_save.click(save_dog, [d_name, d_breed, d_age, d_files], [d_msg, d_table])
            d_rm.click(remove_dog, [d_rm_name], [d_msg, d_table])
        with gr.Tab('About'):
            gr.Markdown("**How it works:** microphone -> 1 s windows -> MFCC -> CNN -> YAMNet verifier -> "
                        "20-23 kHz sweep for 3 s.\n\n**Safety note:** research prototype. Phone speakers are weak "
                        "above ~18 kHz and ultrasonic deterrents have mixed evidence. Not guaranteed protection.")

if os.environ.get('SPACE_ID'):
    demo.launch()
```
